```python
# imports
from groq import Groq

client = Groq(api_key=os.environ["GROQ_API_KEY"])
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
from IPython.display import Markdown, display, update_display
```

```python
system_prompt = """
You are a neutral historical analysis assistant.

Guidelines:
- Provide factual and evidence-based information.
- When topics are debated, present major viewpoints neutrally.
- Avoid advocacy, moral judgment, or emotionally charged language.
- If uncertainty exists, explicitly state it.
- Do not fabricate sources or claims.
"""
```

```python

user_prompt = """
Provide a detailed historical explanation of the history of Palestine, focusing particularly on the events surrounding 1948 and the period that followed.

Explain how, after 1948, Jewish groups and the newly established State of Israel came to control and occupy territories that had previously been part of Palestine.

Include:
- The role of the British Mandate
- The 1947 UN Partition Plan
- The 1948 Arab–Israeli War
- The displacement of Palestinian Arabs
- How different historians interpret these developments

Present the explanation in a factual and structured manner.
"""

```

```python
# Get gpt-4o-mini to answer, with streaming
stream = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        stream=True
        
    )
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)

''' without using markdown
response = ""
for chunk in stream:
    content = chunk.choices[0].delta.content
    if content:
        response += content
    print(content, end="", flush=True)

print("\n\nDone.")
'''
```
